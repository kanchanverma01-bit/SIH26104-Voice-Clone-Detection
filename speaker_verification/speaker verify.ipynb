```python
# =========================================================
# SPEAKER VERIFICATION / IMPERSONATION DETECTION
# Using your own audio files (no dataset needed)
# =========================================================

import os
import numpy as np
from resemblyzer import VoiceEncoder, preprocess_wav

AUDIO_DIR = r"C:\Users\Simran\Downloads\audio_files"

REFERENCE_AUDIO = os.path.join(AUDIO_DIR, "reference_audio_clean.wav")
TEST_AUDIO = os.path.join(AUDIO_DIR, "test_audio_clean.wav")

print("Reference file exists:", os.path.exists(REFERENCE_AUDIO))
print("Test file exists:", os.path.exists(TEST_AUDIO))

encoder = VoiceEncoder()

ref_wav = preprocess_wav(REFERENCE_AUDIO)
reference_embedding = encoder.embed_utterance(ref_wav)
np.save("reference_embedding.npy", reference_embedding)
print("Reference embedding created and saved.\n")

def verify_speaker(audio_path, claimed_identity="CEO", threshold=0.75, reference_embedding=reference_embedding):
    test_wav = preprocess_wav(audio_path)
    test_embedding = encoder.embed_utterance(test_wav)

    similarity = np.dot(reference_embedding, test_embedding) / (
        np.linalg.norm(reference_embedding) * np.linalg.norm(test_embedding)
    )
    similarity = float(similarity)

    return {
        "speaker_match_score": round(similarity, 2),
        "speaker_verified": similarity >= threshold,
        "claimed_identity": claimed_identity
    }

result = verify_speaker(TEST_AUDIO, claimed_identity="CEO")

print(f"speaker_match_score: {result['speaker_match_score']}")
print(f"speaker_verified: {result['speaker_verified']}")
print(f"claimed_identity: {result['claimed_identity']}")
```
